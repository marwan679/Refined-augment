```python
from Refined_Augment import Refined_Augment
import cv2
from Refined_Augment.Diagnostics import Look_for_haarcascade
```

```python
# 1. Initialize the augmenter
augmenter = Refined_Augment()

# 2. Start your webcam
cap = cv2.VideoCapture(0)

# 3. Define the list of assets for your hands
# Hand 1 (usually the first one MediaPipe detects) will get the sword.
# Hand 2 will get the shield.
my_assets = ['wolf.obj', 'AR_photo.png'] 

print("Starting webcam... Press 'q' to quit.")

while True:
    ret, frame = cap.read()
    if not ret:
        print("Failed to grab frame")
        break

    # Flip the frame horizontally for a more natural "mirror" feel
    frame = cv2.flip(frame, 1)

    # 4. Apply the augmentation
    augmented_frame = augmenter.overlay(
        image=frame,
        overlay_paths=my_assets,      # Pass the list of assets here
        target='hand',                # Tell it to look for hands, not faces
        use_mediapipe=True,           # Enable the automatic MediaPipe detector
        position='infront',           # Position the asset directly over the hand
        show_bounding_box=False,      # Set to True if you want to see the tracking boxes
        hand_scale_factor=1.2         # Adjust this to make the objects bigger or smaller
    )

    # 5. Display the final image
    cv2.imshow('Dual Hand Augmentation', augmented_frame)

    # Quit if the user presses 'q'
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Clean up
cap.release()
cv2.destroyAllWindows()
```

```python
import cv2
import matplotlib.pyplot as plt
# (Assume Refined_Augment is defined here)

# 1. Initialize
augmenter = Refined_Augment()
my_assets = ['sword.obj', 'shield.png']

# 2. Load a test image from your hard drive
frame = cv2.imread('AR_photo.png')

# 3. Apply the overlay
augmented_frame = augmenter.overlay(
    image=frame,
    overlay_paths=my_assets,
    target='hand',
    use_mediapipe=True,
    position='infront'
)

# 4. Convert BGR (OpenCV format) to RGB (Matplotlib format)
augmented_frame_rgb = cv2.cvtColor(augmented_frame, cv2.COLOR_BGR2RGB)

# 5. Display right inside the notebook
plt.figure(figsize=(10, 10))
plt.imshow(augmented_frame_rgb)
plt.axis('off')
plt.show()
```
